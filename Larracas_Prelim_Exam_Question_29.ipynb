```python

subjects = []

while True:
    subject_name = input("Enter subject name (or type 'done' to finish): ")
    if subject_name.lower() == 'done':
        break

    while True:
        try:
            marks = float(input(f"Enter marks for {subject_name}: "))
            if 0 <= marks <= 100:
                break
            else:
                print("Invalid marks. Please enter a value between 0 and 100.")
        except ValueError:
            print("Invalid input. Please enter a numeric value for marks.")

    subjects.append({'name': subject_name, 'marks': marks})

def get_grade(marks):
  if marks >= 90:
    return "A+ (PASSED)"
  elif marks >= 80:
    return "A (PASSED)"
  elif marks >= 70:
    return "B+ (PASSED)"
  elif marks >= 60:
    return "B (PASSED)"
  elif marks >= 50:
    return "C (PASSED)"
  elif marks >= 40:
    return "D (PASSED)"
  elif marks < 40:
    return "F (FAIL)"
  else:
    return "Invalid Marks"

def get_performance_remark(marks):
  if marks >= 90:
    return "Outstanding"
  if marks >= 80:
    return "Excellent"
  if marks >= 70:
    return "Very Good"
  if marks >= 60:
    return "Good"
  if marks >= 40:
    return "Satisfactory"
  else:
    return "Needs Improvement"


print("\n--- Results ---\n")
if not subjects:
  print("There are no subjects entered.")
else:
  for subjects in subjects:
    grade = get_grade(subjects['marks'])
    remark = get_performance_remark (subjects['marks'])
    print(f"Subject: {subjects['name']}")
    print(f"Marks: {subjects['marks']:.2f}")
    print(f"Grade: {grade}")
    print(f"Performance: {remark}\n")
```
